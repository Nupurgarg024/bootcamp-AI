```python
#Use DecisionTree to design and train a model for predicting the student performance based on the features we discussed and tried practically. Try to vary the depth (maxdepth=3,4,5) and do a comparative analysis for it's impact on the model's performance.
# Import required libraries
import pandas as pd

# Import function to split the dataset into training and testing data
from sklearn.model_selection import train_test_split

# Import Decision Tree Classifier
from sklearn.tree import DecisionTreeClassifier

# Import accuracy metric
from sklearn.metrics import accuracy_score


# Create student dataset
data = {
    'Study_Hours': [
        2, 3, 5, 6, 7, 8, 4, 9, 1, 10,
        3, 6, 8, 5, 7, 2, 9, 4, 10, 6
    ],

    'Attendance': [
        60, 65, 75, 80, 85, 90, 70, 95, 50, 98,
        68, 82, 92, 78, 88, 55, 94, 72, 97, 84
    ],

    'Previous_Marks': [
        45, 50, 60, 65, 70, 85, 55, 90, 40, 95,
        52, 68, 88, 62, 76, 42, 91, 58, 96, 73
    ],

    'Assignment_Score': [
        40, 45, 55, 65, 70, 85, 50, 90, 35, 95,
        48, 70, 88, 60, 78, 38, 92, 55, 96, 75
    ],

    # Target variable
    'Performance': [
        'Fail', 'Fail', 'Pass', 'Pass', 'Pass',
        'Pass', 'Fail', 'Pass', 'Fail', 'Pass',
        'Fail', 'Pass', 'Pass', 'Pass', 'Pass',
        'Fail', 'Pass', 'Fail', 'Pass', 'Pass'
    ]
}

# Convert the dictionary into a DataFrame
df = pd.DataFrame(data)

# Display the dataset
print("Student Dataset:")
print(df)


# Select input features
X = df[
    [
        'Study_Hours',
        'Attendance',
        'Previous_Marks',
        'Assignment_Score'
    ]
]

# Select target variable
y = df['Performance']


# Split the dataset into training and testing data
# 70% data is used for training
# 30% data is used for testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42
)


# Test different values of maximum tree depth
depths = [3, 4, 5]

print("\nDecision Tree Results:")

for depth in depths:

    # Create Decision Tree model
    model = DecisionTreeClassifier(
        max_depth=depth,
        random_state=42
    )

    # Train the model
    model.fit(X_train, y_train)

    # Make predictions on test data
    y_pred = model.predict(X_test)

    # Calculate accuracy
    accuracy = accuracy_score(y_test, y_pred)

    # Display result
    print("Max Depth:", depth)
    print("Accuracy:", accuracy)
    print()
```

```python
#Write python code to demonstrate application of linear regression in electronics industry.
# Import NumPy for numerical operations
import numpy as np

# Import Matplotlib for plotting
import matplotlib.pyplot as plt

# Import Linear Regression model
from sklearn.linear_model import LinearRegression

# Import evaluation metrics
from sklearn.metrics import mean_squared_error, r2_score


# Temperature values in degree Celsius
# These represent the operating temperature of an electronic device
temperature = np.array([
    20, 25, 30, 35, 40,
    45, 50, 55, 60, 65
]).reshape(-1, 1)


# Power consumption values in Watts
power = np.array([
    50, 55, 61, 66, 72,
    78, 83, 89, 94, 100
])


# Create Linear Regression model
model = LinearRegression()


# Train the model using temperature and power data
model.fit(temperature, power)


# Predict power consumption for the existing temperature values
predicted_power = model.predict(temperature)


# Display the slope of the regression line
print("Slope:", model.coef_[0])


# Display the intercept of the regression line
print("Intercept:", model.intercept_)


# Predict power consumption at a new temperature of 70°C
new_temperature = np.array([[70]])

prediction = model.predict(new_temperature)


# Display predicted power consumption
print("Predicted Power at 70°C:",
      prediction[0])


# Calculate Mean Squared Error
mse = mean_squared_error(
    power,
    predicted_power
)


# Calculate R2 score
r2 = r2_score(
    power,
    predicted_power
)


# Display evaluation results
print("Mean Squared Error:", mse)
print("R2 Score:", r2)


# Plot the original data points
plt.scatter(
    temperature,
    power
)


# Plot the regression line
plt.plot(
    temperature,
    predicted_power
)


# Add labels to the graph
plt.xlabel("Temperature (°C)")
plt.ylabel("Power Consumption (Watts)")

# Add title
plt.title(
    "Temperature vs Power Consumption"
)

# Display the graph
plt.show()
```

```python
#Demonstrate application of unsupervised learning to train a model
# Import pandas for creating and handling the dataset
import pandas as pd

# Import matplotlib for visualization
import matplotlib.pyplot as plt

# Import K-Means clustering algorithm
from sklearn.cluster import KMeans


# Create customer dataset
# Annual spending and number of products purchased
data = {

    'Annual_Spending': [
        10, 12, 15, 18, 20,
        50, 55, 58, 60, 65,
        90, 95, 100, 105, 110
    ],

    'Products_Purchased': [
        2, 3, 2, 4, 3,
        10, 12, 11, 13, 12,
        20, 22, 21, 23, 25
    ]
}


# Convert the dictionary into a DataFrame
df = pd.DataFrame(data)


# Display the original data
print("Customer Data:")
print(df)


# Select the features that will be used for clustering
X = df[
    [
        'Annual_Spending',
        'Products_Purchased'
    ]
]


# Create K-Means model
# We want to divide customers into 3 clusters
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)


# Train the K-Means model
kmeans.fit(X)


# Assign cluster numbers to each customer
df['Cluster'] = kmeans.labels_


# Display the clustering results
print("\nClustering Results:")
print(df)


# Display the center of each cluster
print("\nCluster Centers:")
print(kmeans.cluster_centers_)


# Create scatter plot
plt.scatter(
    df['Annual_Spending'],
    df['Products_Purchased'],
    c=df['Cluster']
)


# Add X-axis label
plt.xlabel("Annual Spending")


# Add Y-axis label
plt.ylabel("Products Purchased")


# Add title
plt.title(
    "Customer Segmentation using K-Means"
)


# Display the graph
plt.show()
```
