```python
#Create a program displaying your skills related to image editing using opencv-python.

import cv2
import matplotlib.pyplot as plt
import numpy as np

# ==========================================================
# STEP 1: READ IMAGE
# ==========================================================

img = cv2.imread("input.jpg")

if img is None:
    print("Error: Image not found!")
else:
    print("Image loaded successfully!")

    # Convert BGR to RGB
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # ======================================================
    # STEP 2: DISPLAY ORIGINAL IMAGE
    # ======================================================

    plt.imshow(img_rgb)
    plt.title("Original Image")
    plt.axis("off")
    plt.show()

    # ======================================================
    # STEP 3: RESIZE IMAGE
    # ======================================================

    resized = cv2.resize(img, (600, 400))

    resized_rgb = cv2.cvtColor(resized, cv2.COLOR_BGR2RGB)

    plt.imshow(resized_rgb)
    plt.title("Resized Image")
    plt.axis("off")
    plt.show()

    # ======================================================
    # STEP 4: ROTATE IMAGE
    # ======================================================

    height, width = resized.shape[:2]
    center = (width // 2, height // 2)

    rotation_matrix = cv2.getRotationMatrix2D(
        center,
        30,
        1.0
    )

    rotated = cv2.warpAffine(
        resized,
        rotation_matrix,
        (width, height)
    )

    rotated_rgb = cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB)

    plt.imshow(rotated_rgb)
    plt.title("Rotated Image")
    plt.axis("off")
    plt.show()

    # ======================================================
    # STEP 5: GRAYSCALE
    # ======================================================

    gray = cv2.cvtColor(
        resized,
        cv2.COLOR_BGR2GRAY
    )

    plt.imshow(gray, cmap="gray")
    plt.title("Grayscale Image")
    plt.axis("off")
    plt.show()

    # ======================================================
    # STEP 6: GAUSSIAN BLUR
    # ======================================================

    blurred = cv2.GaussianBlur(
        resized,
        (5, 5),
        0
    )

    blurred_rgb = cv2.cvtColor(
        blurred,
        cv2.COLOR_BGR2RGB
    )

    plt.imshow(blurred_rgb)
    plt.title("Blurred Image")
    plt.axis("off")
    plt.show()

    # ======================================================
    # STEP 7: EDGE DETECTION
    # ======================================================

    edges = cv2.Canny(
        gray,
        100,
        200
    )

    plt.imshow(edges, cmap="gray")
    plt.title("Edge Detection")
    plt.axis("off")
    plt.show()

    # ======================================================
    # STEP 8: FLIP IMAGE
    # ======================================================

    flipped = cv2.flip(
        resized,
        1
    )

    flipped_rgb = cv2.cvtColor(
        flipped,
        cv2.COLOR_BGR2RGB
    )

    plt.imshow(flipped_rgb)
    plt.title("Flipped Image")
    plt.axis("off")
    plt.show()

    # ======================================================
    # STEP 9: CROP IMAGE
    # ======================================================

    cropped = resized[
        50:350,
        100:500
    ]

    cropped_rgb = cv2.cvtColor(
        cropped,
        cv2.COLOR_BGR2RGB
    )

    plt.imshow(cropped_rgb)
    plt.title("Cropped Image")
    plt.axis("off")
    plt.show()

    # ======================================================
    # STEP 10: BRIGHTNESS ADJUSTMENT
    # ======================================================

    brightness = cv2.convertScaleAbs(
        resized,
        alpha=1.0,
        beta=50
    )

    brightness_rgb = cv2.cvtColor(
        brightness,
        cv2.COLOR_BGR2RGB
    )

    plt.imshow(brightness_rgb)
    plt.title("Brightness Increased")
    plt.axis("off")
    plt.show()

    # ======================================================
    # STEP 11: CONTRAST ADJUSTMENT
    # ======================================================

    contrast = cv2.convertScaleAbs(
        resized,
        alpha=1.5,
        beta=0
    )

    contrast_rgb = cv2.cvtColor(
        contrast,
        cv2.COLOR_BGR2RGB
    )

    plt.imshow(contrast_rgb)
    plt.title("Contrast Adjusted")
    plt.axis("off")
    plt.show()

    # ======================================================
    # STEP 12: SHARPEN IMAGE
    # ======================================================

    sharpening_kernel = np.array([
        [0, -1, 0],
        [-1, 5, -1],
        [0, -1, 0]
    ])

    sharpened = cv2.filter2D(
        resized,
        -1,
        sharpening_kernel
    )

    sharpened_rgb = cv2.cvtColor(
        sharpened,
        cv2.COLOR_BGR2RGB
    )

    plt.imshow(sharpened_rgb)
    plt.title("Sharpened Image")
    plt.axis("off")
    plt.show()

    # ======================================================
    # STEP 13: THRESHOLDING
    # ======================================================

    _, threshold = cv2.threshold(
        gray,
        127,
        255,
        cv2.THRESH_BINARY
    )

    plt.imshow(
        threshold,
        cmap="gray"
    )

    plt.title("Threshold Image")
    plt.axis("off")
    plt.show()

    # ======================================================
    # STEP 14: ADD BORDER
    # ======================================================

    bordered = cv2.copyMakeBorder(
        resized,
        20,
        20,
        20,
        20,
        cv2.BORDER_CONSTANT,
        value=(0, 0, 0)
    )

    bordered_rgb = cv2.cvtColor(
        bordered,
        cv2.COLOR_BGR2RGB
    )

    plt.imshow(bordered_rgb)
    plt.title("Image With Border")
    plt.axis("off")
    plt.show()

    # ======================================================
    # STEP 15: ADD TEXT
    # ======================================================

    edited = resized.copy()

    cv2.putText(
        edited,
        "OpenCV Image Editing",
        (30, 50),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (0, 0, 255),
        2
    )

    # ======================================================
    # STEP 16: DRAW RECTANGLE
    # ======================================================

    cv2.rectangle(
        edited,
        (50, 80),
        (250, 250),
        (255, 0, 0),
        3
    )

    # ======================================================
    # STEP 17: DRAW CIRCLE
    # ======================================================

    cv2.circle(
        edited,
        (450, 150),
        70,
        (0, 255, 0),
        3
    )

    # ======================================================
    # STEP 18: DRAW LINE
    # ======================================================

    cv2.line(
        edited,
        (50, 350),
        (550, 350),
        (255, 255, 0),
        3
    )

    # ======================================================
    # STEP 19: FINAL EDITED IMAGE
    # ======================================================

    edited_rgb = cv2.cvtColor(
        edited,
        cv2.COLOR_BGR2RGB
    )

    plt.imshow(edited_rgb)
    plt.title("Final Edited Image")
    plt.axis("off")
    plt.show()

    # ======================================================
    # STEP 20: SAVE ALL IMPORTANT RESULTS
    # ======================================================

    cv2.imwrite("01_resized.jpg", resized)
    cv2.imwrite("02_rotated.jpg", rotated)
    cv2.imwrite("03_grayscale.jpg", gray)
    cv2.imwrite("04_blurred.jpg", blurred)
    cv2.imwrite("05_edges.jpg", edges)
    cv2.imwrite("06_flipped.jpg", flipped)
    cv2.imwrite("07_cropped.jpg", cropped)
    cv2.imwrite("08_brightness.jpg", brightness)
    cv2.imwrite("09_contrast.jpg", contrast)
    cv2.imwrite("10_sharpened.jpg", sharpened)
    cv2.imwrite("11_threshold.jpg", threshold)
    cv2.imwrite("12_bordered.jpg", bordered)
    cv2.imwrite("13_final_edited.jpg", edited)

    print("-----------------------------------")
    print("IMAGE EDITING COMPLETED")
    print("-----------------------------------")
    print("13 edited images have been saved successfully!")
```
